```python
%%writefile /content/BIT_CD/satquery_detector.py

import os
import sys
import cv2
import torch
import numpy as np

from PIL import Image

# --------------------------------------------------
# BIT PATH
# --------------------------------------------------

BIT_ROOT = "/content/BIT_CD"

if BIT_ROOT not in sys.path:
    sys.path.insert(0, BIT_ROOT)

from datasets.CD_dataset import CDDataAugmentation
from models.basic_model import CDEvaluator


class SatQueryBIT:

    def __init__(
        self,
        checkpoint_name="best_ckpt.pt",
        gpu_id=0,
        img_size=256
    ):

        self.img_size = img_size

        # ------------------------------------------
        # Device
        # ------------------------------------------

        if torch.cuda.is_available():
            self.device = torch.device(f"cuda:{gpu_id}")
        else:
            self.device = torch.device("cpu")

        print("Device:", self.device)

        # ------------------------------------------
        # Create BIT arguments
        # ------------------------------------------

        class Args:
            pass

        args = Args()

        args.gpu_ids = [gpu_id] if torch.cuda.is_available() else []
        args.checkpoint_root = os.path.join(
            BIT_ROOT,
            "checkpoints"
        )

        args.project_name = "BIT_LEVIR"

        args.checkpoint_dir = os.path.join(
            args.checkpoint_root,
            args.project_name
        )

        args.net_G = "base_transformer_pos_s4_dd8_dedim8"

        args.n_class = 2
        args.embed_dim = 256
        args.img_size = img_size

        # ------------------------------------------
        # Load evaluator
        # ------------------------------------------

        self.model = CDEvaluator(args)

        self.model.load_checkpoint(checkpoint_name)

        self.model.eval()

        print("BIT model loaded successfully.")

    # ==================================================
    # LOAD + PREPROCESS IMAGE
    # ==================================================

    def preprocess(self, image_path):

        image = Image.open(image_path).convert("RGB")

        original_size = image.size

        image = np.asarray(image)

        aug = CDDataAugmentation(
            img_size=self.img_size
        )

        [image], _ = aug.transform(
            [image],
            [],
            to_tensor=True
        )

        image = image.unsqueeze(0)

        return image, original_size

    # ==================================================
    # RUN BIT
    # ==================================================

    def predict(self, before_path, after_path):

        print("\nRunning BIT...")
        print("Before:", before_path)
        print("After :", after_path)

        # ------------------------------------------
        # Load images
        # ------------------------------------------

        before, before_size = self.preprocess(
            before_path
        )

        after, after_size = self.preprocess(
            after_path
        )

        # ------------------------------------------
        # Move to GPU/CPU
        # ------------------------------------------

        before = before.to(self.device)
        after = after.to(self.device)

        # ------------------------------------------
        # Create BIT batch
        # ------------------------------------------

        batch = {
            "A": before,
            "B": after
        }

        # ------------------------------------------
        # Run model
        # ------------------------------------------

        with torch.no_grad():

            self.model._forward_pass(batch)

            # BIT internally produces prediction
            prediction = self.model.G_pred

        # ------------------------------------------
        # Convert prediction to numpy
        # ------------------------------------------

        prediction = prediction.detach().cpu().numpy()

        # ------------------------------------------
        # Handle output shape
        # ------------------------------------------

        if prediction.ndim == 4:

            prediction = prediction[0]

            if prediction.shape[0] == 2:
                prediction = np.argmax(
                    prediction,
                    axis=0
                )

            else:
                prediction = prediction[0]

        elif prediction.ndim == 3:

            prediction = prediction[0]

        # ------------------------------------------
        # Binary mask
        # ------------------------------------------

        mask = (
            prediction > 0
        ).astype(np.uint8)

        return {
            "mask": mask,
            "before_size": before_size,
            "after_size": after_size
        }

    # ==================================================
    # ANALYZE MASK
    # ==================================================

    def analyze_mask(self, mask):

        total_pixels = mask.size

        changed_pixels = int(
            mask.sum()
        )

        unchanged_pixels = (
            total_pixels -
            changed_pixels
        )

        change_percentage = (
            changed_pixels /
            total_pixels *
            100
        )

        # ------------------------------------------
        # Connected regions
        # ------------------------------------------

        num_labels, labels, stats, centroids = (
            cv2.connectedComponentsWithStats(
                mask,
                connectivity=8
            )
        )

        regions = []

        for i in range(1, num_labels):

            x = stats[
                i,
                cv2.CC_STAT_LEFT
            ]

            y = stats[
                i,
                cv2.CC_STAT_TOP
            ]

            width = stats[
                i,
                cv2.CC_STAT_WIDTH
            ]

            height = stats[
                i,
                cv2.CC_STAT_HEIGHT
            ]

            area = stats[
                i,
                cv2.CC_STAT_AREA
            ]

            # Ignore tiny noise
            if area < 100:
                continue

            regions.append({
                "x": int(x),
                "y": int(y),
                "width": int(width),
                "height": int(height),
                "area": int(area)
            })

        return {
            "total_pixels": int(total_pixels),
            "changed_pixels": changed_pixels,
            "unchanged_pixels": int(unchanged_pixels),
            "change_percentage": float(
                change_percentage
            ),
            "regions": regions
        }

    # ==================================================
    # CREATE OVERLAY
    # ==================================================

    def create_overlay(
        self,
        after_path,
        mask,
        output_path
    ):

        after = np.array(
            Image.open(after_path).convert("RGB")
        )

        # Resize mask if necessary
        if (
            mask.shape[1],
            mask.shape[0]
        ) != (
            after.shape[1],
            after.shape[0]
        ):

            mask = cv2.resize(
                mask,
                (
                    after.shape[1],
                    after.shape[0]
                ),
                interpolation=cv2.INTER_NEAREST
            )

        overlay = after.copy()

        # Red change regions
        overlay[mask > 0] = [
            255,
            0,
            0
        ]

        # Blend
        result = cv2.addWeighted(
            after,
            0.55,
            overlay,
            0.45,
            0
        )

        Image.fromarray(result).save(
            output_path
        )

        return output_path

    # ==================================================
    # GENERATE TEXT
    # ==================================================

    def generate_text(
        self,
        analysis
    ):

        percentage = analysis[
            "change_percentage"
        ]

        region_count = len(
            analysis["regions"]
        )

        if percentage < 1:
            magnitude = "very little"

        elif percentage < 5:
            magnitude = "a small amount of"

        elif percentage < 15:
            magnitude = "a moderate amount of"

        else:
            magnitude = "a substantial amount of"

        if region_count == 0:

            return (
                "No significant change "
                "was detected between "
                "the two images."
            )

        region_word = (
            "region"
            if region_count == 1
            else "regions"
        )

        return (
            f"BIT detected {magnitude} "
            f"change between the two images. "
            f"Approximately "
            f"{percentage:.1f}% of the image "
            f"area is classified as changed, "
            f"concentrated in "
            f"{region_count} significant "
            f"{region_word}."
        )

    # ==================================================
    # COMPLETE PIPELINE
    # ==================================================

    def analyze(
        self,
        before_path,
        after_path,
        output_dir="/content/BIT_CD/samples/satquery_output"
    ):

        os.makedirs(
            output_dir,
            exist_ok=True
        )

        # ------------------------------------------
        # BIT prediction
        # ------------------------------------------

        result = self.predict(
            before_path,
            after_path
        )

        mask = result["mask"]

        # ------------------------------------------
        # Statistics
        # ------------------------------------------

        analysis = self.analyze_mask(
            mask
        )

        # ------------------------------------------
        # Overlay
        # ------------------------------------------

        overlay_path = os.path.join(
            output_dir,
            "change_overlay.png"
        )

        self.create_overlay(
            after_path,
            mask,
            overlay_path
        )

        # ------------------------------------------
        # Text
        # ------------------------------------------

        text = self.generate_text(
            analysis
        )

        return {
            "mask": mask,
            "overlay_path": overlay_path,
            "statistics": analysis,
            "text": text
        }
```

```python
import cv2
import numpy as np

mask = result["mask"].astype(np.uint8)

# Convert binary mask to 0-255
mask_uint8 = mask * 255

# Connect nearby changed pixels
kernel = np.ones((7, 7), np.uint8)

clean_mask = cv2.morphologyEx(
    mask_uint8,
    cv2.MORPH_CLOSE,
    kernel
)

# Remove small isolated noise
clean_mask = cv2.morphologyEx(
    clean_mask,
    cv2.MORPH_OPEN,
    np.ones((3, 3), np.uint8)
)

# Convert back to binary
clean_mask = (clean_mask > 0).astype(np.uint8)

# Find regions again
num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
    clean_mask,
    connectivity=8
)

MIN_AREA = 200

clean_regions = []

for i in range(1, num_labels):

    x = int(stats[i, cv2.CC_STAT_LEFT])
    y = int(stats[i, cv2.CC_STAT_TOP])
    w = int(stats[i, cv2.CC_STAT_WIDTH])
    h = int(stats[i, cv2.CC_STAT_HEIGHT])
    area = int(stats[i, cv2.CC_STAT_AREA])

    if area < MIN_AREA:
        continue

    clean_regions.append({
        "id": len(clean_regions) + 1,
        "x": x,
        "y": y,
        "width": w,
        "height": h,
        "area": area
    })

clean_regions = sorted(
    clean_regions,
    key=lambda r: r["area"],
    reverse=True
)

print("========== CLEAN REGIONS ==========")
print("Regions:", len(clean_regions))

for r in clean_regions:
    print(
        f"R{r['id']}: "
        f"x={r['x']}, "
        f"y={r['y']}, "
        f"w={r['width']}, "
        f"h={r['height']}, "
        f"area={r['area']}"
    )
```

```python
%%writefile /content/test_region_vlm.py

import torch
from transformers import AutoProcessor, AutoModelForMultimodalLM

MODEL_NAME = "Qwen/Qwen2.5-VL-3B-Instruct"

BEFORE_IMAGE = "/content/satquery_inputs/regions/R3_before.png"
AFTER_IMAGE = "/content/satquery_inputs/regions/R3_after.png"

print("Loading processor...")
processor = AutoProcessor.from_pretrained(MODEL_NAME)

print("Loading model...")
model = AutoModelForMultimodalLM.from_pretrained(
    MODEL_NAME,
    device_map="auto",
    dtype=torch.float16
)

print("Model loaded.")
print("Analyzing region R3...")

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "image",
                "image": BEFORE_IMAGE
            },
            {
                "type": "image",
                "image": AFTER_IMAGE
            },
            {
                "type": "text",
                "text": """
You are analyzing two satellite image crops of the same geographic location.

The first image is BEFORE (T1).
The second image is AFTER (T2).

Compare the two images carefully.

Identify:
1. What changed?
2. Where did the change occur?
3. What type of object or land feature appears to have changed?

Do not invent details that cannot be visually confirmed.
If uncertain, use words such as "appears to", "likely", or "possibly".

Give a concise answer in 2-4 sentences.
"""
            }
        ]
    }
]

inputs = processor.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=150
    )

generated = outputs[0][inputs["input_ids"].shape[-1]:]

answer = processor.decode(
    generated,
    skip_special_tokens=True
)

print("\n==============================")
print("SATQUERY VLM RESULT")
print("==============================")
print(answer)
```
